**Supervised Learning**

In supervised learning, a computer learns from data that is already labeled. This means the data has both an input and a correct output. The goal is for the computer to figure out the rules that connect the input to the output. Think of it like a student learning with a teacher who gives them the right answers.

**Models**

**Classification:** This is used when the answer is a specific category. For example, telling a computer to look at pictures and sort them into "cat" or "dog."

**Regression:** This is used when the answer is a number. For example, using a person's age and location to predict their house price.

**Unsupervised Learning:**
In unsupervised learning, the computer is given data that is not labeled. There are no right or wrong answers provided. The computer has to find patterns and hidden structures in the data on its own. It's like a student exploring a new topic and finding connections without any help.

**Models**

**Clustering:** This groups similar data points together. For example, looking at customer information and grouping people with similar shopping habits.

**Dimensionality Reduction:** This simplifies complex data by reducing the number of features or variables. This can make the data easier to work with and understand. For example, simplifying a list of 50 different health measurements into just a few main health categories.

**Supervised Learning Models 🧑‍🏫**
These models learn from labeled data to make predictions.

**Linear Regression:** Predicts a number, like house prices.

**Logistic Regression:** Predicts a category, like if an email is spam or not.

**Decision Trees:** Uses a tree-like structure to make decisions based on data.

**Random Forest:** An improved version of decision trees that uses many trees to get a better result.

**Support Vector Machines (SVM):** Finds the best line to separate different categories of data.

**k-Nearest Neighbors (k-NN):** Predicts by looking at the closest data points.

**Neural Networks:** Inspired by the human brain, these are used for complex tasks like image recognition.

**Unsupervised Learning Models 🕵️‍♀️**
These models find patterns in unlabeled data.

**K-Means Clustering:** Groups data points into a specific number of clusters.

**DBSCAN:** Finds clusters of data based on how close data points are to each other.

**Principal Component Analysis (PCA):** Reduces complex data into simpler forms while keeping important information.

**Independent Component Analysis (ICA):** Separates a mixed signal into its original parts.

**Apriori Algorithm:** Finds rules about items that often appear together, like items in a shopping cart.

# **Linear Regression**

```python
import numpy as np

class LinearRegressionScratch:
    def __init__(self, lr=0.01, n_iters=1000):
        self.lr = lr
        self.n_iters = n_iters
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0

        for _ in range(self.n_iters):
            y_predicted = np.dot(X, self.weights) + self.bias
            dw = (1 / n_samples) * np.dot(X.T, (y_predicted - y))
            db = (1 / n_samples) * np.sum(y_predicted - y)

            self.weights -= self.lr * dw
            self.bias -= self.lr * db

    def predict(self, X):
        return np.dot(X, self.weights) + self.bias

# Example usage:
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([3, 5, 7, 9, 11])  # Linear relation: y = 2x + 1

model = LinearRegressionScratch(lr=0.01, n_iters=1000)
model.fit(X, y)
print("Weights:", model.weights)
print("Bias:", model.bias)
print("Predictions:", model.predict(np.array([[6], [7]])))
```

# **Logistic Regression**

```python
import numpy as np

class LogisticRegressionScratch:
    def __init__(self, lr=0.01, n_iters=1000):
        self.lr = lr
        self.n_iters = n_iters
        self.weights = None
        self.bias = None

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0

        for _ in range(self.n_iters):
            linear_model = np.dot(X, self.weights) + self.bias
            y_predicted = self.sigmoid(linear_model)

            dw = (1 / n_samples) * np.dot(X.T, (y_predicted - y))
            db = (1 / n_samples) * np.sum(y_predicted - y)

            self.weights -= self.lr * dw
            self.bias -= self.lr * db

    def predict_proba(self, X):
        linear_model = np.dot(X, self.weights) + self.bias
        return self.sigmoid(linear_model)

    def predict(self, X):
        y_predicted_probs = self.predict_proba(X)
        return [1 if i >= 0.5 else 0 for i in y_predicted_probs]

# Example usage
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])  # Labels

model = LogisticRegressionScratch(lr=0.1, n_iters=1000)
model.fit(X, y)

print("Weights:", model.weights)
print("Bias:", model.bias)
print("Predictions:", model.predict(np.array([[1.5], [3.5], [5]])))
```

# **Decision Trees**

```python
import numpy as np

class DecisionTreeNode:
    def __init__(self, feature=None, threshold=None, left=None, right=None, *, value=None):
        self.feature = feature
        self.threshold = threshold
        self.left = left
        self.right = right
        self.value = value

    def is_leaf_node(self):
        return self.value is not None

class DecisionTreeScratch:
    def __init__(self, max_depth=5, min_samples_split=2):
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.root = None

    def fit(self, X, y):
        self.root = self._grow_tree(X, y)

    def _grow_tree(self, X, y, depth=0):
        n_samples, n_features = X.shape
        n_labels = len(np.unique(y))

        # Stopping conditions
        if (depth >= self.max_depth or n_labels == 1 or n_samples < self.min_samples_split):
            leaf_value = self._most_common_label(y)
            return DecisionTreeNode(value=leaf_value)

        # Find best split
        best_feature, best_thresh = self._best_split(X, y, n_features)

        # Split dataset
        left_idxs = X[:, best_feature] <= best_thresh
        right_idxs = X[:, best_feature] > best_thresh
        left = self._grow_tree(X[left_idxs], y[left_idxs], depth+1)
        right = self._grow_tree(X[right_idxs], y[right_idxs], depth+1)
        return DecisionTreeNode(best_feature, best_thresh, left, right)

    def _best_split(self, X, y, n_features):
        best_gain = -1
        split_idx, split_thresh = None, None
        for feature in range(n_features):
            thresholds = np.unique(X[:, feature])
            for t in thresholds:
                gain = self._information_gain(y, X[:, feature], t)
                if gain > best_gain:
                    best_gain = gain
                    split_idx = feature
                    split_thresh = t
        return split_idx, split_thresh

    def _information_gain(self, y, X_column, threshold):
        # Parent entropy
        parent_entropy = self._entropy(y)

        # Generate splits
        left_idxs = X_column <= threshold
        right_idxs = X_column > threshold
        if len(y[left_idxs]) == 0 or len(y[right_idxs]) == 0:
            return 0

        # Weighted average child entropy
        n = len(y)
        n_l, n_r = len(y[left_idxs]), len(y[right_idxs])
        e_l, e_r = self._entropy(y[left_idxs]), self._entropy(y[right_idxs])
        child_entropy = (n_l / n) * e_l + (n_r / n) * e_r

        # Information gain
        return parent_entropy - child_entropy

    def _entropy(self, y):
        hist = np.bincount(y)
        ps = hist / len(y)
        return -np.sum([p * np.log2(p) for p in ps if p > 0])

    def _most_common_label(self, y):
        counts = np.bincount(y)
        return np.argmax(counts)

    def predict(self, X):
        return np.array([self._traverse_tree(x, self.root) for x in X])

    def _traverse_tree(self, x, node):
        if node.is_leaf_node():
            return node.value
        if x[node.feature] <= node.threshold:
            return self._traverse_tree(x, node.left)
        return self._traverse_tree(x, node.right)

# Example usage
X = np.array([[2], [3], [10], [19], [20]])
y = np.array([0, 0, 1, 1, 1])

tree = DecisionTreeScratch(max_depth=3)
tree.fit(X, y)
print("Predictions:", tree.predict(np.array([[5], [15]])))
```

# **Random Forest**

```python
import numpy as np
from collections import Counter

class DecisionTreeScratch:
    def __init__(self, max_depth=5):
        self.max_depth = max_depth
        self.tree = None

    def fit(self, X, y, depth=0):
        n_samples, n_features = X.shape
        if depth == self.max_depth or len(set(y)) == 1:
            self.tree = Counter(y).most_common(1)[0][0]
            return

        # choose random feature
        feat_idx = np.random.randint(0, n_features)
        threshold = np.median(X[:, feat_idx])

        left_mask = X[:, feat_idx] <= threshold
        right_mask = X[:, feat_idx] > threshold

        if sum(left_mask) == 0 or sum(right_mask) == 0:
            self.tree = Counter(y).most_common(1)[0][0]
            return

        self.tree = {
            'feature': feat_idx,
            'threshold': threshold,
            'left': DecisionTreeScratch(self.max_depth),
            'right': DecisionTreeScratch(self.max_depth)
        }
        self.tree['left'].fit(X[left_mask], y[left_mask], depth+1)
        self.tree['right'].fit(X[right_mask], y[right_mask], depth+1)

    def predict_one(self, x):
        if not isinstance(self.tree, dict):
            return self.tree
        if x[self.tree['feature']] <= self.tree['threshold']:
            return self.tree['left'].predict_one(x)
        else:
            return self.tree['right'].predict_one(x)

    def predict(self, X):
        return np.array([self.predict_one(x) for x in X])

class RandomForestScratch:
    def __init__(self, n_estimators=5, max_depth=5):
        self.n_estimators = n_estimators
        self.max_depth = max_depth
        self.trees = []

    def fit(self, X, y):
        self.trees = []
        for _ in range(self.n_estimators):
            idxs = np.random.choice(len(X), len(X), replace=True)
            X_sample, y_sample = X[idxs], y[idxs]
            tree = DecisionTreeScratch(max_depth=self.max_depth)
            tree.fit(X_sample, y_sample)
            self.trees.append(tree)

    def predict(self, X):
        predictions = np.array([tree.predict(X) for tree in self.trees])
        predictions = np.swapaxes(predictions, 0, 1)
        return np.array([Counter(row).most_common(1)[0][0] for row in predictions])

# Example
X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

rf = RandomForestScratch(n_estimators=5, max_depth=3)
rf.fit(X, y)
print("Predictions:", rf.predict(np.array([[2], [5]])))
```

# **Support Vector Machines (SVM):**

```python
import numpy as np

class LinearSVM:
    def __init__(self, learning_rate=0.001, lambda_param=0.01, n_iters=1000):
        self.lr = learning_rate
        self.lambda_param = lambda_param
        self.n_iters = n_iters
        self.w = None
        self.b = None

    def fit(self, X, y):
        n_samples, n_features = X.shape
        y_ = np.where(y <= 0, -1, 1)  # convert labels to -1, +1

        self.w = np.zeros(n_features)
        self.b = 0

        for _ in range(self.n_iters):
            for idx, x_i in enumerate(X):
                condition = y_[idx] * (np.dot(x_i, self.w) - self.b) >= 1
                if condition:
                    self.w -= self.lr * (2 * self.lambda_param * self.w)
                else:
                    self.w -= self.lr * (2 * self.lambda_param * self.w - np.dot(x_i, y_[idx]))
                    self.b -= self.lr * y_[idx]

    def predict(self, X):
        approx = np.dot(X, self.w) - self.b
        return np.sign(approx)

# Example
X = np.array([[1,2],[2,3],[3,3],[2,1],[3,2]])
y = np.array([0,0,0,1,1])

svm = LinearSVM()
svm.fit(X, y)
print("Predictions:", svm.predict(X))
```

# **k-Nearest Neighbors (k-NN):**

```python
import numpy as np
from collections import Counter

class KNN:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def predict(self, X):
        return [self._predict(x) for x in X]

    def _predict(self, x):
        # compute distances
        distances = [np.linalg.norm(x - x_train) for x_train in self.X_train]
        k_idx = np.argsort(distances)[:self.k]
        k_neighbor_labels = [self.y_train[i] for i in k_idx]
        most_common = Counter(k_neighbor_labels).most_common(1)
        return most_common[0][0]

# Example
X_train = np.array([[1,2],[2,3],[3,3],[6,5],[7,7]])
y_train = np.array([0,0,0,1,1])

knn = KNN(k=3)
knn.fit(X_train, y_train)
print("Prediction:", knn.predict([[5,5]]))
```

# **Neural Networks:**

```python
import numpy as np

class SimpleNN:
    def __init__(self, input_size, hidden_size, output_size, lr=0.01):
        self.lr = lr
        # Weights initialization
        self.W1 = np.random.randn(input_size, hidden_size)
        self.b1 = np.zeros((1, hidden_size))
        self.W2 = np.random.randn(hidden_size, output_size)
        self.b2 = np.zeros((1, output_size))

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))

    def sigmoid_derivative(self, x):
        return x * (1 - x)

    def fit(self, X, y, epochs=1000):
        for _ in range(epochs):
            # Forward pass
            z1 = np.dot(X, self.W1) + self.b1
            a1 = self.sigmoid(z1)
            z2 = np.dot(a1, self.W2) + self.b2
            a2 = self.sigmoid(z2)

            # Backpropagation
            error = y - a2
            d_a2 = error * self.sigmoid_derivative(a2)

            error_hidden = d_a2.dot(self.W2.T)
            d_a1 = error_hidden * self.sigmoid_derivative(a1)

            # Update weights
            self.W2 += a1.T.dot(d_a2) * self.lr
            self.b2 += np.sum(d_a2, axis=0, keepdims=True) * self.lr
            self.W1 += X.T.dot(d_a1) * self.lr
            self.b1 += np.sum(d_a1, axis=0, keepdims=True) * self.lr

    def predict(self, X):
        z1 = np.dot(X, self.W1) + self.b1
        a1 = self.sigmoid(z1)
        z2 = np.dot(a1, self.W2) + self.b2
        a2 = self.sigmoid(z2)
        return np.round(a2)

# Example (XOR)
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

nn = SimpleNN(input_size=2, hidden_size=2, output_size=1, lr=0.1)
nn.fit(X, y, epochs=5000)
print("Predictions:", nn.predict(X))
```

# **K-Means Clustering:**

```python
import numpy as np

class KMeans:
    def __init__(self, K=2, max_iters=100):
        self.K = K
        self.max_iters = max_iters
        self.centroids = None
        self.clusters = None

    def fit(self, X):
        n_samples, n_features = X.shape
        random_idx = np.random.choice(n_samples, self.K, replace=False)
        self.centroids = X[random_idx]

        for _ in range(self.max_iters):
            self.clusters = self._create_clusters(X)
            centroids_old = self.centroids
            self.centroids = self._calculate_centroids(X)

            if np.all(centroids_old == self.centroids):
                break

    def _create_clusters(self, X):
        clusters = [[] for _ in range(self.K)]
        for idx, point in enumerate(X):
            centroid_idx = self._closest_centroid(point)
            clusters[centroid_idx].append(idx)
        return clusters

    def _closest_centroid(self, point):
        distances = [np.linalg.norm(point - centroid) for centroid in self.centroids]
        return np.argmin(distances)

    def _calculate_centroids(self, X):
        centroids = np.zeros((self.K, X.shape[1]))
        for cluster_idx, cluster in enumerate(self.clusters):
            cluster_mean = np.mean(X[cluster], axis=0)
            centroids[cluster_idx] = cluster_mean
        return centroids

    def predict(self, X):
        return [self._closest_centroid(point) for point in X]

# Example
X = np.array([[1,2],[1,4],[1,0],[10,2],[10,4],[10,0]])
kmeans = KMeans(K=2)
kmeans.fit(X)
print("Centroids:", kmeans.centroids)
print("Predictions:", kmeans.predict(X))
```

# **DBSCAN:**

```python
import numpy as np
from collections import deque

class DBSCAN:
    def __init__(self, eps=1.0, min_samples=5):
        self.eps = eps
        self.min_samples = min_samples
        self.labels = None

    def fit(self, X):
        n_samples = X.shape[0]
        self.labels = np.full(n_samples, -1)  # -1 = noise
        cluster_id = 0

        for i in range(n_samples):
            if self.labels[i] != -1:
                continue

            neighbors = self._region_query(X, i)
            if len(neighbors) < self.min_samples:
                self.labels[i] = -1  # noise
            else:
                self._expand_cluster(X, i, neighbors, cluster_id)
                cluster_id += 1

    def _expand_cluster(self, X, point_idx, neighbors, cluster_id):
        self.labels[point_idx] = cluster_id
        queue = deque(neighbors)

        while queue:
            neighbor_idx = queue.popleft()
            if self.labels[neighbor_idx] == -1:
                self.labels[neighbor_idx] = cluster_id
            if self.labels[neighbor_idx] != -1:
                continue
            self.labels[neighbor_idx] = cluster_id

            new_neighbors = self._region_query(X, neighbor_idx)
            if len(new_neighbors) >= self.min_samples:
                queue.extend(new_neighbors)

    def _region_query(self, X, idx):
        distances = np.linalg.norm(X - X[idx], axis=1)
        return np.where(distances <= self.eps)[0]

# Example
X = np.array([[1,2],[2,2],[2,3],[8,7],[8,8],[25,80]])
db = DBSCAN(eps=2, min_samples=2)
db.fit(X)
print("Cluster Labels:", db.labels)
```

# **Apriori Algorithm:**

```python
from itertools import combinations

def get_support(transactions, itemset):
    count = 0
    for transaction in transactions:
        if itemset.issubset(transaction):
            count += 1
    return count / len(transactions)

def apriori(transactions, min_support, min_confidence):
    items = set()
    for t in transactions:
        items.update(t)
    items = list(items)

    # Generate frequent 1-itemsets
    freq_itemsets = []
    support_data = {}

    for item in items:
        sup = get_support(transactions, {item})
        if sup >= min_support:
            freq_itemsets.append({item})
            support_data[frozenset({item})] = sup

    k = 2
    final_itemsets = freq_itemsets.copy()

    while freq_itemsets:
        candidates = []
        for i in range(len(freq_itemsets)):
            for j in range(i+1, len(freq_itemsets)):
                union = freq_itemsets[i] | freq_itemsets[j]
                if len(union) == k and union not in candidates:
                    candidates.append(union)

        freq_itemsets = []
        for c in candidates:
            sup = get_support(transactions, c)
            if sup >= min_support:
                freq_itemsets.append(c)
                support_data[frozenset(c)] = sup

        final_itemsets.extend(freq_itemsets)
        k += 1

    # Generate rules
    rules = []
    for itemset in final_itemsets:
        if len(itemset) > 1:
            for A in itemset:
                Aset = {A}
                Bset = itemset - Aset
                conf = support_data[frozenset(itemset)] / support_data[frozenset(Aset)]
                if conf >= min_confidence:
                    rules.append((Aset, Bset, conf))

    return final_itemsets, rules

# Example transactions
transactions = [
    {"milk", "bread", "eggs"},
    {"milk", "bread"},
    {"milk", "eggs"},
    {"bread", "eggs"},
    {"milk", "bread", "eggs"}
]

itemsets, rules = apriori(transactions, min_support=0.4, min_confidence=0.7)
print("Frequent Itemsets:", itemsets)
print("Association Rules:", rules)
```

**Using Library**

```python
from mlxtend.frequent_patterns import apriori, association_rules
import pandas as pd

# Example transactions
dataset = [
    ["milk", "bread", "eggs"],
    ["milk", "bread"],
    ["milk", "eggs"],
    ["bread", "eggs"],
    ["milk", "bread", "eggs"]
]

# Convert to one-hot encoded DataFrame
df = pd.DataFrame([{item: (item in transaction) for item in {"milk","bread","eggs"}} for transaction in dataset])

# Generate frequent itemsets
frequent_itemsets = apriori(df, min_support=0.4, use_colnames=True)

# Generate rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

print(frequent_itemsets)
print(rules)
```
